### NSES datatype corrigeren

from sting to float64

```python
def fix_nses(value):
    """Herstelt NSES-waarden waarin het decimaalteken vervangen is door duizendtal-punten."""
    if pd.isna(value):
        return np.nan

    value = str(value)
    sign, digits = ("-", value[1:]) if value.startswith("-") else ("", value)
    digits = digits.replace(".", "")

    return float(f"{sign}{digits[0]}.{digits[1:]}")


df_filtered["NSES"] = df_filtered["NSES"].apply(fix_nses)
```

```python
datasets = {
    "df_trainA": df_trainA,
    "df_testA": df_testA,
    "df_trainB": df_trainB,
    "df_testB": df_testB,
    "df_trainC": df_trainC,
    "df_testC" : df_testC,
}

for name, data in datasets.items():
    missing_cols = data.columns[data.isna().any()]

    for col in missing_cols:
        data[f"{col}_MISSING"] = data[col].isna().astype(int)

    print(f"{name}: {len(missing_cols)} missing-flag kolommen toegevoegd")
```

```python
#Imputation 
mode_columns = [
    "DIAG_FIBROMYALGIA_ACR",
    "DIAG_CFS_CDC",
    "DIAG_IBS_ROME3",
    "SPORTS_T1",
    "SMOKING",
    "HTN_MED_T1",
    "EDUCATION_LOWER_T1",
    "EDUCATION_LOWER_T2",
    "RESPIRATORY_DISEASE_T1",
]

# Mode wordt berekend op de train set en toegepast op train + bijbehorende test set (voorkomt data leakage)
for col in mode_columns:
    mode_a = df_trainA[col].mode()[0]
    df_trainA[col] = df_trainA[col].fillna(mode_a)
    df_testA[col] = df_testA[col].fillna(mode_a)

    mode_b = df_trainB[col].mode()[0]
    df_trainB[col] = df_trainB[col].fillna(mode_b)
    df_testB[col] = df_testB[col].fillna(mode_b)
    
    mode_c = df_trainC[col].mode()[0]
    df_trainC[col] = df_trainC[col].fillna(mode_c)
    df_testC[col] = df_testC[col].fillna(mode_c)
```

```python
median_columns = [
    "LLDS",
    "NSES",
    "MWK_VAL",
    "BKR_T1",
    "CHO_T1",
    "HDC_T1",
    "LDC_T1",
    "TGL_T1",
    "HBF_T1",
    "MAP_T1",
    "BMI_T2",
    "HIP_T2",
    "WAIST_T2",
    "BKR_T2",
    "CHO_T2",
    "HDC_T2",
    "LDC_T2",
    "TGL_T2",
    "HBF_T2",
    "MAP_T2",
    "CHO_T3",
    "HDC_T3",
    "LDC_T3",
    "TGL_T3",
    "HIP_T3",
    "WAIST_T3",
]

# Mediaan wordt berekend op de train set en toegepast op train + bijbehorende test set (voorkomt data leakage)
for col in median_columns:
    median_a = df_trainA[col].median()
    df_trainA[col] = df_trainA[col].fillna(median_a)
    df_testA[col] = df_testA[col].fillna(median_a)

    median_b = df_trainB[col].median()
    df_trainB[col] = df_trainB[col].fillna(median_b)
    df_testB[col] = df_testB[col].fillna(median_b)
    
    median_c = df_trainC[col].median()
    df_trainC[col] = df_trainC[col].fillna(median_c)
    df_testC[col] = df_testC[col].fillna(median_c)
```

### Engeneer features na imputatie

```python
#BRI calculatie op basis van middelomtrek en lengte, daarna WAIST en HEIGHT droppen om multicollineariteit te voorkomen
datasets = {
    "df_trainA": df_trainA,
    "df_testA": df_testA,
    "df_trainB": df_trainB,
    "df_testB": df_testB,
    "df_trainC": df_trainC,
    "df_testC" : df_testC,
}

columns_to_drop_bri = [
    "HEIGHT_T1", "HEIGHT_T2", "HEIGHT_T3",
    "WAIST_T1", "WAIST_T2", "WAIST_T3",
]

for name, data in datasets.items():
    for t in ["T1", "T2", "T3"]:
        wc = data[f"WAIST_{t}"]
        height = data[f"HEIGHT_{t}"]
        data[f"BRI_{t}"] = 364.2 - 365.5 * np.sqrt(1 - (wc / (2 * np.pi)) ** 2 / (0.5 * height) ** 2)

    data.drop(columns=columns_to_drop_bri, inplace=True)
```

```python
#non-HDL cholesterol calculation based on total cholesterol, HDL 
datasets = {
    "df_trainA": df_trainA,
    "df_testA": df_testA,
    "df_trainB": df_trainB,
    "df_testB": df_testB,
    "df_trainC": df_trainC,
    "df_testC" : df_testC,
}

for name, data in datasets.items():
    for t in ["T1", "T2", "T3"]:
        data[f"NHDC_{t}"] = data[f"CHO_{t}"] - data[f"HDC_{t}"]

    data.drop(columns=["CHO_T1", "CHO_T2", "CHO_T3"], inplace=True)


```

```python
#calculation of the TGL/HDL ratio
datasets = {
    "df_trainA": df_trainA,
    "df_testA": df_testA,
    "df_trainB": df_trainB,
    "df_testB": df_testB,
    "df_trainC": df_trainC,
    "df_testC" : df_testC,
}

for name, data in datasets.items():
    for t in ["T1", "T2", "T3"]:
        data[f"THR_{t}"] = data[f"TGL_{t}"] / data[f"HDC_{t}"]

    data.drop(columns=["TGL_T1", "TGL_T2", "TGL_T3", "HDC_T1", "HDC_T2", "HDC_T3"], inplace=True)
```
