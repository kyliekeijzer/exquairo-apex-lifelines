### NSES datatype corrigeren

from sting to float64

```python
def fix_nses(value):
    """Herstelt NSES-waarden waarin het decimaalteken vervangen is door duizendtal-punten."""
    if pd.isna(value):
        return np.nan

    value = str(value)
    sign, digits = ("-", value[1:]) if value.startswith("-") else ("", value)
    digits = digits.replace(".", "")

    return float(f"{sign}{digits[0]}.{digits[1:]}")


df_filtered["NSES"] = df_filtered["NSES"].apply(fix_nses)
```

### Handling missings part II - imputeren

Zet missing-flags (berekend vóór imputatie)

Diagnoses (~10–15% NaN)


DIAG_FIBROMYALGIA_ACR, DIAG_CFS_CDC, DIAG_IBS_ROME3
→ Mode (of één blok-flag als alle drie missing) + missing-indicator per kolom.
(Geen NaN→0 tenzij bootcamp “leeg=nee” bevestigt; mode is praktischer baseline.)

Scores


Kolom	Methode
LLDS	Median (op train) + flag
NSES	Median + flag (parse al oké in xlsx)
MWK_VAL, MWK_NO_VAL, SCOR_VAL, SCOR_NO_VAL	Median + flag (vaak samen missing)
Binaries (licht)
SMOKING, SPORTS_T1, HTN_MED_T1, EDUCATION_LOWER_T1, EDUCATION_LOWER_T2, RESPIRATORY_DISEASE_T1
→ Mode + flag
(EDUCATION_LOWER_T2: alleen houden als je T2 mag gebruiken; anders droppen i.v.m. leakage)

Labs / anthro / vitals


Groep	Methode
T1: BKR/CHO/HDC/LDC/TGL, HBF_T1, MAP_T1	Median + flag (T1-labs: 63 rijen vaak samen → of die rijen droppen i.p.v. imputeren)
T2: labs, BMI/HIP/WAIST, HBF/MAP	Median + flag; niet als feature voor HBAC_T1
T3: labs, HIP/WAIST	Idem; check HIP_T3 ≤0 → missing

```python
#Overzicht van missing data per dataset (train/test)
datasets = {
    "df_trainA": df_trainA,
    "df_testA": df_testA,
    "df_trainB": df_trainB,
    "df_testB": df_testB,
}

for name, data in datasets.items():
    missing = data.isna().sum()
    missing = missing[missing > 0].sort_values(ascending=False)

    print(f"{name}:")
    print(missing)
    print()
```

### Flag colom for missing values 

```python
datasets = {
    "df_trainA": df_trainA,
    "df_testA": df_testA,
    "df_trainB": df_trainB,
    "df_testB": df_testB,
}

for name, data in datasets.items():
    missing_cols = data.columns[data.isna().any()]

    for col in missing_cols:
        data[f"{col}_MISSING"] = data[col].isna().astype(int)

    print(f"{name}: {len(missing_cols)} missing-flag kolommen toegevoegd")
```

```python
#Imputation 
mode_columns = [
    "DIAG_FIBROMYALGIA_ACR",
    "DIAG_CFS_CDC",
    "DIAG_IBS_ROME3",
    "SPORTS_T1",
    "SMOKING",
    "HTN_MED_T1",
    "EDUCATION_LOWER_T1",
    "EDUCATION_LOWER_T2",
    "RESPIRATORY_DISEASE_T1",
]

# Mode wordt berekend op de train set en toegepast op train + bijbehorende test set (voorkomt data leakage)
for col in mode_columns:
    mode_a = df_trainA[col].mode()[0]
    df_trainA[col] = df_trainA[col].fillna(mode_a)
    df_testA[col] = df_testA[col].fillna(mode_a)

    mode_b = df_trainB[col].mode()[0]
    df_trainB[col] = df_trainB[col].fillna(mode_b)
    df_testB[col] = df_testB[col].fillna(mode_b)
```

```python
median_columns = [
    "LLDS",
    "NSES",
    "MWK_VAL",
    "BKR_T1",
    "CHO_T1",
    "HDC_T1",
    "LDC_T1",
    "TGL_T1",
    "HBF_T1",
    "MAP_T1",
    "BMI_T2",
    "HIP_T2",
    "WAIST_T2",
    "BKR_T2",
    "CHO_T2",
    "HDC_T2",
    "LDC_T2",
    "TGL_T2",
    "HBF_T2",
    "MAP_T2",
    "CHO_T3",
    "HDC_T3",
    "LDC_T3",
    "TGL_T3",
    "HIP_T3",
    "WAIST_T3",
]

# Mediaan wordt berekend op de train set en toegepast op train + bijbehorende test set (voorkomt data leakage)
for col in median_columns:
    median_a = df_trainA[col].median()
    df_trainA[col] = df_trainA[col].fillna(median_a)
    df_testA[col] = df_testA[col].fillna(median_a)

    median_b = df_trainB[col].median()
    df_trainB[col] = df_trainB[col].fillna(median_b)
    df_testB[col] = df_testB[col].fillna(median_b)
```
